## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My research lane is a web-based Inventory and Sales Management System for small businesses such as sari-sari stores. The website helps business owners record sales, monitor inventory levels, and export reports to Excel for business analysis.

The machine learning task is **Classification**.

The model predicts whether a product needs to be restocked based on inventory levels and historical sales data. The output supports inventory management by helping store owners decide which products should be reordered before they run out of stock.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target variable is **needs_restock**.

This target is based on an observed outcome from historical inventory and sales records. Products that reached a critical stock level or sold out before being replenished are labeled as needing restock.

Target values:

- **1 = Needs Restock**
- **0 = Does Not Need Restock**

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The success metric for this classification task is **F1-score**.

F1-score balances Precision and Recall, making it suitable because both false positives and false negatives can negatively affect a business.

A good model should correctly identify products that require restocking while minimizing unnecessary purchase recommendations.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The **unit of analysis** is a single product.

Each row in the dataset represents a unique product and its relevant attributes at a specific point in time. These attributes include current stock levels, recent sales data, and the target variable indicating whether the product needs to be restocked.

```python
import pandas as pd

data = {
    "product_name": [
        "Rice",
        "Instant Noodles",
        "Coffee",
        "Soft Drinks",
        "Soap"
    ],
    "current_stock": [5, 30, 8, 50, 12],
    "units_sold_last_7_days": [42, 12, 30, 15, 6],
    "average_daily_sales": [6, 2, 4, 2, 1],
    "days_since_restock": [10, 4, 8, 5, 12],
    "needs_restock": [1, 0, 1, 0, 0]
}

df = pd.DataFrame(data)

display(df)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "Restock when stock is below 10" does not work well because products sell at different rates.

For example, five bags of rice may sell out in one day, while five bottles of soy sauce may last for several weeks.

Machine learning considers multiple factors at the same time, including:

- Current inventory level
- Historical sales
- Average daily sales
- Sales trends
- Time since the last restock

By learning patterns from historical data, the model can make smarter restocking recommendations than a fixed rule. This helps reduce stock shortages, avoid overstocking, and improve inventory management.
