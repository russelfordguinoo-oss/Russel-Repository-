## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

# Simulate a starter CSV for demonstration purposes
# In a real scenario, you would load your actual CSV using pd.read_csv('path/to/your/data.csv')
num_days = 30
dates = pd.date_range(start='2023-10-01', periods=num_days)

data = {
    'Date': dates,
    'Hours_Focused': np.random.uniform(2, 8, num_days).round(1),
    'Tasks_Completed': np.random.randint(3, 15, num_days),
    'Well_being_Score': np.random.uniform(5, 10, num_days).round(1) # Scale of 1-10
}

df = pd.DataFrame(data)

print("Simulated Data Head:")
display(df.head())
```

```python
# Display 2-3 real numbers relevant to the research lane

# 1. Average Daily Focused Hours
average_hours_focused = df['Hours_Focused'].mean()
print(f"1. Average daily focused hours: {average_hours_focused:.2f}")

# 2. Average Daily Tasks Completed
average_tasks_completed = df['Tasks_Completed'].mean()
print(f"2. Average daily tasks completed: {average_tasks_completed:.2f}")

# 3. Average Well-being Score
average_well_being = df['Well_being_Score'].mean()
print(f"3. Average well-being score (1-10 scale): {average_well_being:.2f}")

# You can also show min/max to illustrate variability or potential for improvement
min_well_being = df['Well_being_Score'].min()
max_well_being = df['Well_being_Score'].max()
print(f"Well-being score range: {min_well_being} (min) to {max_well_being} (max)")
```
