```python
# Show columns related to our research question
for col in df.columns:
    if any(word in col.lower() for word in [
        "click", "session", "engag", "trend", "30d"
    ]):
        print(col)
```

```python
print("90th percentile clicks:",
      df["clicks_90d"].quantile(0.90))

print("90th percentile sessions:",
      df["sessions_90d"].quantile(0.90))
```

```python
print(df["trend_pct"].describe())
```

```python
print(df["trend_direction"].value_counts())
```

```python
strong_pages = df[
    (
        (df["clicks_90d"] >= df["clicks_90d"].quantile(0.90))
        |
        (df["sessions_90d"] >= df["sessions_90d"].quantile(0.90))
    )
    &
    (df["trend_direction"] == "stable")
]

print("Strong benchmark pages:", len(strong_pages))
print("Percentage of all pages:", len(strong_pages) / len(df) * 100)
```

```python
print(
    strong_pages[
        [
            "content_id",
            "clicks_90d",
            "sessions_90d",
            "trend_direction",
            "trend_pct"
        ]
    ].head(10)
)
```

```python
other_pages = df.loc[~df["content_id"].isin(strong_pages["content_id"])]

comparison = pd.DataFrame({
    "Strong benchmark pages": [
        strong_pages["clicks_90d"].mean(),
        strong_pages["sessions_90d"].mean(),
        strong_pages["trend_pct"].mean()
    ],
    "Other pages": [
        other_pages["clicks_90d"].mean(),
        other_pages["sessions_90d"].mean(),
        other_pages["trend_pct"].mean()
    ]
}, index=["Average clicks (90d)", "Average sessions (90d)", "Average trend %"])

print(comparison)
```

The starter dataset contains 3,000 pages. Using my initial benchmark rule — top 10% in clicks or sessions and a stable trend — 152 pages (5.1%) qualify as strong benchmark pages. This gives me a manageable group of pages to study for common patterns in clicks, sessions, and performance trend stability.
